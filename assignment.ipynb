**Question 4:** Write a Python program to find the largest of three numbers.

```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
num3 = float(input("Enter the third number: "))

if num1 >= num2 and num1 >= num3:
    largest = num1
elif num2 >= num1 and num2 >= num3:
    largest = num2
else:
    largest = num3

print("The largest number is:", largest)
```

**Question 5:** Write a Python program to calculate the factorial of a number.

```python
num = int(input("Enter a number: "))
factorial = 1

if num < 0:
    print("Factorial is not possible for a negative number")
else:
    for i in range(1, num + 1):
        factorial = factorial * i

    print("The factorial of", num, "is", factorial)
```

**Question 16:** Write a Python program to find the sum of the first n natural numbers.

```python
n = int(input("Enter a number: "))
total = 0

for i in range(1, n + 1):
    total = total + i

print("The sum of first", n, "natural numbers is", total)
```

**Question 17:** Write a Python program to check if a year is a leap year.

```python
year = int(input("Enter a year: "))

if year % 400 == 0:
    print(year, "is a leap year")
elif year % 100 == 0:
    print(year, "is not a leap year")
elif year % 4 == 0:
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

**Question 18:** Write a Python program to reverse a string.

```python
word = input("Enter a word or sentence: ")
reversed_word = ""

for letter in word:
    reversed_word = letter + reversed_word

print("Reversed text:", reversed_word)
```

**Question 19:** Write a Python program to check if a string is a palindrome.

```python
word = input("Enter a word: ")
reversed_word = word[::-1]

if word == reversed_word:
    print(word, "is a palindrome")
else:
    print(word, "is not a palindrome")
```

**Question 20:** Write a Python program to sort a list of numbers in ascending order.

```python
values = input("Enter numbers separated by spaces: ")

numbers = []
for value in values.split():
    numbers.append(int(value))

numbers.sort()
print("Numbers in ascending order:", numbers)
```
